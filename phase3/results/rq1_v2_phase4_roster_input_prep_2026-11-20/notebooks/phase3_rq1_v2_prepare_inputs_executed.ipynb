# Phase 3 RQ1 v2 — prepare inputs

This step freezes XNLI samples for English (`en`), German (`de`), Spanish (`es`), French (`fr`), Swahili (`sw`), Arabic (`ar`), Urdu (`ur`), Bulgarian (`bg`), Russian (`ru`), and Hindi (`hi`). It prepares URIEL+ and sentence-embedding priors. Turn Kaggle Internet on. Embedding inference is explicitly set to CPU; no model is fine-tuned here. A Hugging Face token is not required for the public XNLI data or embedding model.

On success, verify the printed summary has 10 languages, 5 orders, 3 seeds, checksums for the URIEL+ and embedding matrices, and a data-manifest hash. The published-interference matrix uses the frozen Phase 4 k=5 URIEL+ imputation for Arabic; its neighbors, source checksum, and output checksum are recorded. The output ZIP contains the protocol, prior matrices, and data hash—not XNLI sentence text. Keep it in a private Kaggle dataset and attach it to the training and analysis notebooks. Grambank is excluded if its coverage report shows missing languages.

Before running: push the repository changes to GitHub so this notebook can clone them.

```python
from pathlib import Path
import subprocess
import sys

REPO_URL = "https://github.com/Mani-S-2709/asymmetric-interference-continual-learning..git"
REPO_ROOT = Path("/kaggle/working/asymmetric-interference-continual-learning")
if not REPO_ROOT.exists():
    subprocess.run(["git", "clone", "--depth", "1", REPO_URL, str(REPO_ROOT)], check=True)
elif (REPO_ROOT / ".git").is_dir():
    subprocess.run(["git", "-C", str(REPO_ROOT), "pull", "--ff-only"], check=True)
else:
    raise FileExistsError(f"Repository path exists but is not a Git clone: {REPO_ROOT}")

subprocess.run([
    sys.executable, "-m", "pip", "install", "-q",
    "datasets==5.0.1", "urielplus==1.3.3", "sentence-transformers",
], check=True)
sys.path.insert(0, str(REPO_ROOT))

OUTPUT_DIR = Path("/kaggle/working/phase3_rq1_v2_inputs")
command = [
    sys.executable, "-m", "phase3.rq1_v2_workflow", "prepare-inputs",
    "--protocol", str(REPO_ROOT / "phase3" / "configs" / "rq1_v2_10lang_protocol.json"),
    "--output-dir", str(OUTPUT_DIR),
    "--embedding-device", "cpu",
]
command.append("--impute-arabic-interference")
interference_sources = list(Path("/kaggle/input").rglob("interference_matrix.csv"))
if len(interference_sources) > 1:
    raise ValueError("Attach at most one published interference source matrix.")
if interference_sources:
    command.extend(["--interference-source-matrix", str(interference_sources[0])])
    print("Using the attached published matrix and the frozen k=5 Arabic imputation.")
else:
    print("Using the repository's published matrix and the frozen k=5 Arabic imputation.")
subprocess.run(command, cwd=REPO_ROOT, check=True)

INPUT_ZIP = OUTPUT_DIR / "phase3_rq1_v2_inputs.zip"
if not INPUT_ZIP.is_file():
    raise FileNotFoundError(f"Input bundle was not created: {INPUT_ZIP}")
print(f"Create a private Kaggle dataset from: {INPUT_ZIP}")
```
